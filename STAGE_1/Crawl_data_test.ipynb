```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from bs4 import BeautifulSoup
import pandas as pd
import time
from urllib.parse import urljoin

BASE_URL = "https://www.cgv.vn"

options = Options()
options.add_argument("--start-maximized")

driver = webdriver.Chrome(options=options)

try:
    driver.get(BASE_URL)

    time.sleep(5)

    soup = BeautifulSoup(driver.page_source, "html.parser")

    movies = soup.select("li.poster-film")

    print("Số phim:", len(movies))

    data = []

    for movie in movies:

        img = movie.select_one("img.product-img")
        rating = movie.select_one(".movie-rating")
        detail_link = movie.select_one('a[title="Xem chi tiết"]')

        movie_name = img.get("alt") if img else None

        image_url = img.get("src") if img else None

        movie_rating = (
            rating.get_text(strip=True)
            if rating else None
        )

        movie_url = (
            urljoin(BASE_URL, detail_link.get("href"))
            if detail_link else None
        )

        data.append({
            "MovieName": movie_name,
            "MovieURL": movie_url,
            "ImageURL": image_url,
            "Rating": movie_rating
        })

    df = pd.DataFrame(data)

    print(df)

    df.to_csv(
        "cgv_movies.csv",
        index=False,
        encoding="utf-8-sig"
    )

finally:
    driver.quit()
```

```python
from selenium import webdriver
from bs4 import BeautifulSoup
import pandas as pd
from urllib.parse import urljoin
import time

BASE_URL = "https://www.cgv.vn/"

driver = webdriver.Chrome()

try:
    driver.get(BASE_URL)

    # Chờ trang tải
    time.sleep(5)

    # Lấy HTML sau khi Chrome tải trang
    soup = BeautifulSoup(driver.page_source, "html.parser")

    # Tìm các thẻ phim
    movies = soup.select("li.poster-film")

    print("Số lượng phim tìm thấy:", len(movies))

    data = []

    for movie in movies:

        # Hình ảnh
        img = movie.select_one("img.product-img")

        # Rating
        rating = movie.select_one(".nmovie-rating")

        # Link chi tiết
        detail_link = movie.select_one('a[title="Xem chi tiết"]')

        # Tên phim
        movie_name = img.get("alt") if img else None

        # URL hình ảnh
        image_url = img.get("src") if img else None

        # Rating
        movie_rating = (
            rating.get_text(strip=True)
            if rating else None
        )

        # URL phim
        movie_url = (
            urljoin(BASE_URL, detail_link.get("href"))
            if detail_link else None
        )

        data.append({
            "MovieName": movie_name,
            "MovieURL": movie_url,
            "ImageURL": image_url,
            "Rating": movie_rating
        })

    # Chuyển sang DataFrame
    df = pd.DataFrame(data)

    print(df.head(10))

finally:
    driver.quit()
```

```python
from selenium import webdriver
from bs4 import BeautifulSoup
import pandas as pd
import time

driver = webdriver.Chrome()

url = "https://www.cgv.vn/default/forgotten-island.html"

try:
    driver.get(url)
    time.sleep(3)

    soup = BeautifulSoup(driver.page_source, "html.parser")

    data = {
        "MovieURL": url
    }

    # Tên phim
    title = soup.select_one("h1")

    if title:
        data["MovieName"] = title.get_text(" ", strip=True)

    # Các thông tin chi tiết
    movie_info = soup.select("div.movie-info")

    for item in movie_info:

        label = item.select_one("label")
        value = item.select_one(".std")

        if label and value:

            field = label.get_text(" ", strip=True)
            value_text = value.get_text(" ", strip=True)

            if "Đạo diễn" in field:
                data["Director"] = value_text

            elif "Diễn viên" in field:
                data["Actors"] = value_text

            elif "Thể loại" in field:
                data["Genre"] = value_text

            elif "Khởi chiếu" in field:
                data["ReleaseDate"] = value_text

            elif "Thời lượng" in field:
                data["Duration"] = value_text

            elif "Ngôn ngữ" in field:
                data["Language"] = value_text

    print(data)

finally:
    driver.quit()
```

```python
from selenium import webdriver
from bs4 import BeautifulSoup
import pandas as pd
import time

driver = webdriver.Chrome()

results = []

try:

    # =========================
    # BƯỚC 1: LẤY DANH SÁCH PHIM
    # =========================

    driver.get("https://www.cgv.vn/")

    time.sleep(3)

    soup = BeautifulSoup(driver.page_source, "html.parser")

    # Loại cycle-sentinel để tránh phim bị lặp
    movies = soup.select(
        "li.poster-film:not(.cycle-sentinel)"
    )

    print("Số phim:", len(movies))

    movie_list = []

    for movie in movies:

        img = movie.select_one("img.product-img")

        detail_link = movie.select_one(
            'a[title="Xem chi tiết"]'
        )

        rating = movie.select_one(
            ".nmovie-rating"
        )

        if not img or not detail_link:
            continue

        movie_name = img.get("alt")

        movie_url = detail_link.get("href")

        movie_rating = (
            rating.get_text(" ", strip=True)
            if rating else None
        )

        movie_list.append({
            "MovieName": movie_name,
            "MovieURL": movie_url,
            "Rating": movie_rating
        })


    # =========================
    # BƯỚC 2: VÀO TỪNG PHIM
    # =========================

    for i, movie in enumerate(movie_list, start=1):

        print(
            f"[{i}/{len(movie_list)}] "
            f"{movie['MovieName']}"
        )

        driver.get(movie["MovieURL"])

        time.sleep(2)

        soup = BeautifulSoup(
            driver.page_source,
            "html.parser"
        )

        data = {
            "MovieName": movie["MovieName"],
            "MovieURL": movie["MovieURL"],
            "Rating": movie["Rating"],
            "Director": None,
            "Actors": None,
            "Genre": None,
            "ReleaseDate": None,
            "Duration": None,
            "Language": None
        }

        # =========================
        # LẤY THÔNG TIN movie-info
        # =========================

        movie_info = soup.select(
            "div.movie-info"
        )

        for item in movie_info:

            label = item.select_one("label")
            value = item.select_one(".std")

            if not label or not value:
                continue

            field = label.get_text(
                " ",
                strip=True
            )

            value_text = value.get_text(
                " ",
                strip=True
            )

            if "Đạo diễn" in field:
                data["Director"] = value_text

            elif "Diễn viên" in field:
                data["Actors"] = value_text

            elif "Thể loại" in field:
                data["Genre"] = value_text

            elif "Khởi chiếu" in field:
                data["ReleaseDate"] = value_text

            elif "Thời lượng" in field:
                data["Duration"] = value_text

            elif "Ngôn ngữ" in field:
                data["Language"] = value_text


        results.append(data)


finally:
    driver.quit()


# =========================
# BƯỚC 3: DATAFRAME
# =========================

df = pd.DataFrame(results)

print("\n===== KẾT QUẢ =====")

print(df.head(10))

print("\nSố dòng:", len(df))

print("\nSố cột:", len(df.columns))


# =========================
# BƯỚC 4: LƯU FILE
# =========================

df.to_csv(
    "cgv_movie_details.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\nĐã lưu: cgv_movie_details.csv")
```
